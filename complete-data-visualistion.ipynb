# **Top 5 Value**

```python
top_pub = data['Publisher'].value_counts().index[:5]
top_pub
```

**Top 5 Genre**

```python
top_genre=data['Genre'].value_counts().index[:5]
top_genre
```

**Top 5 platforms**

```python
top_platforms=data['Platform'].value_counts().index[:5]
top_platforms
```

```python
top_5_data = data.loc[(data['Publisher'].isin(top_pub)) &(data['Genre'].isin(top_genre)) &(data['Platform'].isin(top_platforms))]

top_5_data
```

```python
plt.figure(figsize=(12,6))

sns.countplot(data=top_5_data, x='Publisher', hue='Platform')

plt.xticks(rotation=90, fontsize=10)

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(data=top_5_data, x='Publisher', hue='Genre')

plt.xticks(rotation=90, fontsize=10)

plt.show()
```

```python
stack_data = pd.crosstab(index=top_5_data['Publisher'], columns=top_5_data['Platform'])
stack_data

```

```python

stack_data.plot(kind='bar', stacked=True)
plt.show()
```
